Data Preprocessing and Label Encoding

```python
# Step 1: Import necessary libraries
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.utils import to_categorical

# Step 2: Inspect the dataset
print("Dataset Columns:", combined_df.columns)
print("Dataset Info:")
combined_df.info()

# Step 3: Preprocess the data
# Rename columns to match the dataset's structure
combined_df.columns = ['timestamp', 'back_x', 'back_y', 'back_z', 'thigh_x', 'thigh_y', 'thigh_z', 'label']

# Handle missing values
combined_df.dropna(inplace=True)

# Encode activity labels
label_encoder = LabelEncoder()
combined_df['label'] = label_encoder.fit_transform(combined_df['label'])

# Separate features and labels
X = combined_df[['back_x', 'back_y', 'back_z', 'thigh_x', 'thigh_y', 'thigh_z']].values
y = combined_df['label'].values

# Normalize the features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Step 4: Split the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)

# Step 5: Build the neural network model
model = Sequential([
    Dense(64, activation='relu', input_dim=X_train.shape[1]),
    Dense(32, activation='relu'),
    Dense(len(label_encoder.classes_), activation='softmax')
])

# Compile the model
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

# Step 6: Train the model
model.fit(X_train, y_train, epochs=10, batch_size=32, validation_data=(X_test, y_test))

# Step 7: Evaluate the model
loss, accuracy = model.evaluate(X_test, y_test)
print(f"Test Accuracy: {accuracy * 100:.2f}%")
```

Reshape Features into Sequences for LSTM Input

```python
import numpy as np

# Assuming X is your original NumPy array with shape (1832560, 6)
sequence_length = 128
n_features = 6

# Calculate the number of complete sequences
n_samples = X.shape[0] // sequence_length

# Trim the array to have only full sequences
X_trimmed = X[:n_samples * sequence_length]

# Reshape the trimmed array
X_reshaped = X_trimmed.reshape((n_samples, sequence_length, n_features))
```

Reshape and Prepare Labels for Sequences

```python
# Assuming y is your original label array
y_trimmed = y[:n_samples * sequence_length]
y_reshaped = y_trimmed.reshape((n_samples, sequence_length))
y_seq = y_reshaped[:, -1]  # Taking the label of the last timestep in each sequence
```

Train-Test Split for Sequence Data

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_reshaped, y_seq, test_size=0.2, random_state=42)
```

Define LSTM Model Architecture

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense

model = Sequential()
model.add(LSTM(64, input_shape=(sequence_length, n_features)))
model.add(Dense(32, activation='relu'))
model.add(Dense(len(np.unique(y_seq)), activation='softmax'))  # Assuming classification task
```

Compile and Train LSTM Model

```python
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])

model.fit(X_train, y_train, epochs=10, batch_size=32, validation_data=(X_test, y_test))
```

Evaluate Model Accuracy on Test Data

```python
loss, accuracy = model.evaluate(X_test, y_test)
print(f"Test Accuracy: {accuracy * 100:.2f}%")
```

 Fine-Tune the Model

```python
# Install keras-tuner
!pip install -q -U keras-tuner

# Imports
from kerastuner.tuners import RandomSearch
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dropout, Dense, Input
from tensorflow.keras.optimizers import Adam
from sklearn.preprocessing import LabelEncoder
import numpy as np

# === Dummy data for example (replace with your actual data) ===
X_train = np.random.rand(100, 10, 6)  # 100 samples, 10 timesteps, 6 features
X_test = np.random.rand(20, 10, 6)

y_train_raw = np.random.choice(['walk', 'sit', 'stand'], 100)
y_test_raw = np.random.choice(['walk', 'sit', 'stand'], 20)

# === Label Encoding ===
label_encoder = LabelEncoder()
y_train = label_encoder.fit_transform(y_train_raw)
y_test = label_encoder.transform(y_test_raw)

# === Build Model Function ===
def build_model(hp):
    model = Sequential()
    model.add(Input(shape=(10, 6)))  # Input shape based on your time steps and features

    model.add(LSTM(units=hp.Int('units_1', min_value=32, max_value=128, step=32),
                   activation='relu',
                   return_sequences=True))
    model.add(Dropout(hp.Float('dropout_1', 0.1, 0.5, step=0.1)))

    model.add(LSTM(units=hp.Int('units_2', min_value=16, max_value=64, step=16),
                   activation='relu'))
    model.add(Dropout(hp.Float('dropout_2', 0.1, 0.5, step=0.1)))

    model.add(Dense(len(label_encoder.classes_), activation='softmax'))
    model.compile(
        optimizer=Adam(hp.Choice('learning_rate', [1e-2, 1e-3, 1e-4])),
        loss='sparse_categorical_crossentropy',
        metrics=['accuracy'])
    return model

# === Keras Tuner Random Search ===
tuner = RandomSearch(
    build_model,
    objective='val_accuracy',
    max_trials=2,
    executions_per_trial=1,
    directory='har_tuning',
    project_name='har_lstm')

# === Start Tuning ===
tuner.search(X_train, y_train, epochs=10, validation_data=(X_test, y_test))

# === Get Best Model ===
best_model = tuner.get_best_models(num_models=1)[0]

# === Evaluate Best Model ===
loss, accuracy = best_model.evaluate(X_test, y_test)
print(f"Test Accuracy: {accuracy:.2f}")
```

Real-Time Inference Script

```python
import numpy as np
import time
import csv
import matplotlib.pyplot as plt

# Parameters
window_size = 10  # same as training time-step
features = 6      # number of features per timestep
num_windows = 5
delay_between_windows = 1  # seconds

# For live plotting
predictions_list = []
window_ids = []

# Create CSV to log predictions
log_file = "real_time_predictions.csv"
with open(log_file, mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(["Timestamp", "Window", "Predicted Activity", "Confidence"])

print("Starting Real-Time Inference...\n")

# Ensure X_test_reshaped is defined or reshape X_test if needed
if 'X_test_reshaped' not in locals():  # Check if it's not defined already
    if len(X_test.shape) == 2:  # If X_test is 2D, reshape it to (samples, 10, 6)
        X_test_reshaped = X_test.reshape((X_test.shape[0], 10, 6))
    else:  # If already 3D
        X_test_reshaped = X_test
    print(f"Reshaped X_test to: {X_test_reshaped.shape}")

# Real-time loop for activity prediction
for i in range(num_windows):
    if i >= len(X_test_reshaped):
        break

    # Get one reshaped window
    sample_window = X_test_reshaped[i].reshape(1, window_size, features)

    # Model prediction
    pred_class = best_model.predict(sample_window, verbose=0)
    predicted_label = np.argmax(pred_class)
    activity_label = label_encoder.inverse_transform([predicted_label])[0]
    confidence = np.max(pred_class)

    # Print prediction
    print(f"[Window {i+1}] Predicted Activity: {activity_label} (Confidence: {confidence:.2f})")

    # Save prediction to CSV
    with open(log_file, mode='a', newline='') as file:
        writer = csv.writer(file)
        writer.writerow([time.strftime("%Y-%m-%d %H:%M:%S"), i+1, activity_label, f"{confidence:.2f}"])

    # Store predictions for live plotting
    predictions_list.append(activity_label)
    window_ids.append(i+1)

    # Live plot
    plt.clf()
    plt.title("Live Activity Prediction")
    plt.xlabel("Window")
    plt.ylabel("Predicted Activity")
    plt.plot(window_ids, predictions_list, marker='o', linestyle='-', color='b')
    plt.xticks(window_ids)
    plt.pause(0.5)

    # Simulate delay between windows
    time.sleep(delay_between_windows)

print("\nReal-Time Inference Completed.")

# Display final plot with all predictions
plt.show()
```
